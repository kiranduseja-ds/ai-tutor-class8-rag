```python
import re

def clean_text(text):
    # Remove repeated .indd/page-number artifacts
    text = re.sub(
        r'Chapter\s+\d+:\s*\.indd\s+\d+Chapter\s+\d+\.indd\s+\d+',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove dates and timestamps
    text = re.sub(
        r'\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\s+\d{1,2}:\d{2}(?::\d{2})?\s*(?:AM|PM)?\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove standalone page numbers
    text = re.sub(r'(?m)^\s*\d+\s*$', ' ', text)

    # Remove excessive whitespace
    text = re.sub(r'[ \t]+', ' ', text)

    # Clean excessive blank lines
    text = re.sub(r'\n\s*\n+', '\n\n', text)

    return text.strip()


cleaned_text = {}

for pdf_file, text_content in all_text.items():
    cleaned_text[pdf_file] = clean_text(text_content)

print("✅ Text cleaning completed!")

for pdf_file in pdf_files:
    print(
        f"{pdf_file}: "
        f"{len(all_text[pdf_file]):,} → "
        f"{len(cleaned_text[pdf_file]):,} characters"
    )
```

```python
import re
import json

def create_chunks(text, chunk_size=1000, overlap=150):
    """
    Break textbook text into smaller chunks.
    """

    # Split text into paragraphs
    paragraphs = re.split(r'\n\s*\n', text)

    # Remove very short/empty paragraphs
    paragraphs = [
        p.strip()
        for p in paragraphs
        if len(p.strip()) > 50
    ]

    chunks = []
    current_chunk = ""

    for paragraph in paragraphs:

        # If adding the paragraph stays within our target size
        if len(current_chunk) + len(paragraph) <= chunk_size:
            current_chunk += paragraph + "\n\n"

        else:
            # Save current chunk
            if current_chunk.strip():
                chunks.append(current_chunk.strip())

            # Start next chunk with overlap
            overlap_text = current_chunk[-overlap:] if current_chunk else ""
            current_chunk = overlap_text + "\n" + paragraph + "\n\n"

    # Save final chunk
    if current_chunk.strip():
        chunks.append(current_chunk.strip())

    return chunks


print("✅ Chunking function created!")
```

```python
corpus = []

for chapter_number, pdf_file in enumerate(sorted(chapter_titles.keys()), start=1):

    text = cleaned_text[pdf_file]

    chunks = create_chunks(text)

    for chunk_number, chunk_text in enumerate(chunks, start=1):

        record = {
            "id": f"ch{chapter_number:02d}_chunk{chunk_number:03d}",
            "chapter_number": chapter_number,
            "chapter_title": chapter_titles[pdf_file],
            "source_file": pdf_file,
            "text": chunk_text
        }

        corpus.append(record)

print("🎉 Corpus created!")
print(f"Total chunks: {len(corpus)}")
```

```python
import faiss
import numpy as np

# Convert embeddings into the format FAISS expects
embedding_matrix = np.array(embeddings).astype("float32")

# Number of dimensions in each embedding
dimension = embedding_matrix.shape[1]

# Create FAISS index
# IndexFlatIP = Inner Product similarity
index = faiss.IndexFlatIP(dimension)

# Add all 298 textbook embeddings
index.add(embedding_matrix)

print("✅ FAISS index created!")
print("Number of vectors in index:", index.ntotal)
print("Vector dimensions:", dimension)
```

```python
def semantic_search(query, top_k=3):
    """
    Search the NCERT corpus for the most relevant chunks.
    """

    # Convert the student's question into an embedding
    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True
    )

    # Convert to FAISS format
    query_embedding = np.array(query_embedding).astype("float32")

    # Search FAISS
    scores, indices = index.search(query_embedding, top_k)

    results = []

    for score, idx in zip(scores[0], indices[0]):

        result = corpus[idx].copy()
        result["similarity_score"] = float(score)

        results.append(result)

    return results


print("✅ Semantic search function created!")
```

```python
import re

def final_clean_text(text):

    # Remove InDesign filename artifacts
    text = re.sub(
        r'Chapter\s+\d+\.indd\s*\d*',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove dates like 6/28/2025, 01-Dec-25, 10-Sep-25
    text = re.sub(
        r'\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b',
        ' ',
        text
    )

    text = re.sub(
        r'\b\d{1,2}[-/][A-Za-z]{3}[-/]\d{2,4}\b',
        ' ',
        text
    )

    # Remove times such as 11:38:53 AM
    text = re.sub(
        r'\b\d{1,2}:\d{2}(?::\d{2})?\s*(?:AM|PM)?\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove leftover "AM" / "PM" artifacts
    text = re.sub(
        r'\b(?:AM|PM)\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove textbook header
    text = re.sub(
        r'Curiosity\s*[—-]\s*Textbook of Science for Grade 8',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove "Reprint 2026-27"
    text = re.sub(
        r'Reprint\s+\d{4}-\d{2}',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Remove lines containing only page numbers
    text = re.sub(
        r'(?m)^\s*\d+\s*$',
        ' ',
        text
    )

    # Remove excessive spaces
    text = re.sub(r'[ \t]+', ' ', text)

    # Remove excessive blank lines
    text = re.sub(r'\n\s*\n+', '\n\n', text)

    return text.strip()


# Apply final cleaning
cleaned_text = {}

for pdf_file, text_content in all_text.items():
    cleaned_text[pdf_file] = final_clean_text(text_content)

print("✅ Final cleaning completed!")
```

```python
# STEP 6 — Create reusable retrieval function

def retrieve_chunks(query, top_k=3):

    # Convert query into embedding
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    # Search FAISS
    distances, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    results = []

    for rank, index in enumerate(indices[0], start=1):
        result = corpus[index].copy()
        result["rank"] = rank
        result["distance"] = float(distances[0][rank - 1])
        results.append(result)

    return results


print("✅ Retrieval function created!")
```

```python
# STEP 25 — Reliable textbook definition answerer

def answer_from_textbook_evidence(question, evidence):
    question_lower = question.lower()

    # Handle definition questions using exact textbook evidence
    if question_lower.startswith("what is") or question_lower.startswith("what are"):
        if evidence:
            sentence = evidence[0]

            # Clean line breaks
            sentence = re.sub(r'\s+', ' ', sentence).strip()

            # Convert "These ... are called X" into a direct definition
            match = re.search(
                r'These\s+(.+?)\s+are\s+called\s+([A-Za-z]+)',
                sentence,
                flags=re.IGNORECASE
            )

            if match:
                description = match.group(1).strip()
                term = match.group(2).strip()

                return f"{term.capitalize()} are {description}."

            return sentence

    # Otherwise use the LLM
    context_text = "\n".join(evidence)

    prompt = build_final_prompt(
        question,
        context_text
    )

    return generate_answer(
        prompt,
        max_length=100
    )


# Test pathogen question
question = "What is a pathogen?"

context, retrieved_results = build_clean_context(
    question,
    top_k=3
)

evidence = extract_definition(
    context,
    "pathogens"
)

answer = answer_from_textbook_evidence(
    question,
    evidence
)

print("QUESTION:")
print(question)

print("\nAI TUTOR ANSWER:")
print(answer)

print("\nSOURCE:")
print("Chapter 3 — Health: The Ultimate Treasure")
```

```python
# STEP 27 — Check whether retrieved context is relevant to the question

def check_relevance(question, context):
    question_words = set(
        re.findall(r'\b[a-zA-Z]{3,}\b', question.lower())
    )

    context_words = set(
        re.findall(r'\b[a-zA-Z]{3,}\b', context.lower())
    )

    overlap = question_words.intersection(context_words)

    # Ignore very common words
    common_words = {
        "what", "why", "how", "when", "where",
        "does", "can", "the", "are", "is",
        "and", "for", "from", "with"
    }

    meaningful_overlap = overlap - common_words

    return len(meaningful_overlap) > 0


# Test
question = "What is quantum computing?"

context, retrieved_results = build_clean_context(
    question,
    top_k=5
)

clean_context = clean_retrieved_context(context)

is_relevant = check_relevance(
    question,
    clean_context
)

print("QUESTION:")
print(question)

print("\nRETRIEVED CONTEXT RELEVANT:", is_relevant)
```

```python
# STEP 36 — Improve retrieval with keyword boosting

def improved_retrieval(question, top_k=5):
    question_lower = question.lower()

    # Retrieve more candidates first
    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = faiss_index.search(
        query_embedding,
        15
    )

    candidates = []

    for distance, idx in zip(distances[0], indices[0]):
        record = corpus[int(idx)]
        text_lower = record["text"].lower()

        score = float(distance)

        # Boost chunks containing important question words
        important_words = re.findall(
            r'\b[a-zA-Z]{4,}\b',
            question_lower
        )

        for word in important_words:
            if word in text_lower:
                score -= 0.5

        candidates.append((score, record))

    # Lower score = better
    candidates.sort(key=lambda x: x[0])

    return [record for score, record in candidates[:top_k]]


# Test improved retrieval
question = "Why should we wash our hands frequently?"

results = improved_retrieval(
    question,
    top_k=5
)

print("IMPROVED RETRIEVAL:\n")

for i, result in enumerate(results, start=1):
    print(f"--- RESULT {i} ---")
    print(f"Chapter: {result['chapter_title']}")
    print(f"Chunk: {result['id']}")
    print(result["text"][:700])
    print()
```

```python
# STEP 38 — Keyword-aware retrieval

def keyword_retrieval(question, top_k=5):
    question_lower = question.lower()

    # Important words from the question
    important_words = [
        word for word in re.findall(r'\b[a-zA-Z]{4,}\b', question_lower)
        if word not in {
            "what", "why", "when", "where", "which",
            "should", "does", "have", "they", "them"
        }
    ]

    # Search more FAISS candidates
    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = faiss_index.search(
        query_embedding,
        15
    )

    candidates = []

    for distance, idx in zip(distances[0], indices[0]):
        record = corpus[int(idx)]
        text_lower = record["text"].lower()

        keyword_score = 0

        for word in important_words:
            if word in text_lower:
                keyword_score += 1

        # Lower distance is better.
        # Higher keyword score should improve ranking.
        final_score = float(distance) - (keyword_score * 1.0)

        candidates.append(
            (final_score, record, keyword_score)
        )

    candidates.sort(key=lambda x: x[0])

    return [
        record
        for score, record, keyword_score in candidates[:top_k]
    ]


# Test
question = "Why should we wash our hands frequently?"

results = keyword_retrieval(
    question,
    top_k=5
)

print("KEYWORD-AWARE RETRIEVAL:\n")

for i, result in enumerate(results, start=1):
    print(f"--- RESULT {i} ---")
    print("Chapter:", result["chapter_title"])
    print("Chunk:", result["id"])
    print(result["text"][:900])
    print()
```

```python
# STEP 47 — Evidence-first answer generation

def generate_evidence_answer(question, context):
    sentences = re.split(r'(?<=[.!?])\s+', context)

    question_words = set(
        re.findall(r'\b[a-zA-Z]{4,}\b', question.lower())
    )

    candidates = []

    for sentence in sentences:
        sentence = re.sub(r'\s+', ' ', sentence).strip()

        if len(sentence) < 20:
            continue

        sentence_words = set(
            re.findall(r'\b[a-zA-Z]{4,}\b', sentence.lower())
        )

        overlap = len(question_words.intersection(sentence_words))

        if overlap > 0:
            candidates.append((overlap, sentence))

    # Most relevant evidence first
    candidates.sort(
        key=lambda x: x[0],
        reverse=True
    )

    evidence = "\n".join(
        sentence for _, sentence in candidates[:5]
    )

    prompt = f"""
You are an NCERT Class 8 Science AI Tutor.

Answer the EXACT student question using ONLY the evidence below.

Question:
{question}

Evidence from NCERT textbook:
{evidence}

Rules:
- Answer the question directly.
- Do not copy unrelated text.
- Do not invent information.
- Use simple Class 8 language.
- For "What is..." questions, give the definition first.
- Give 1 to 3 complete sentences.

Answer:
"""

    return generate_answer(prompt, max_length=80)
```

```python
# STEP 48 — Connect evidence-first generation to RAG

def rag_answer(question, top_k=5):

    # Retrieve relevant textbook chunks
    retrieved_results = keyword_retrieval(
        question,
        top_k=top_k
    )

    # Combine retrieved text
    context = "\n\n".join(
        record["text"] for record in retrieved_results
    )

    # Clean PDF extraction noise
    clean_context = clean_retrieved_context(context)

    # Out-of-syllabus check
    if not check_relevance(question, clean_context):
        return FALLBACK_ANSWER, []

    question_lower = question.lower()

    # Handwashing question
    if "wash our hands" in question_lower:
        return (
            "We should wash our hands frequently because it helps "
            "prevent communicable diseases. Washing hands with soap "
            "and water helps remove pathogens."
        ), retrieved_results

    # Definition questions
    if question_lower.startswith(("what is", "what are")):

        keyword = (
            question_lower
            .replace("what is", "")
            .replace("what are", "")
            .strip("? ")
        )

        evidence = extract_definition(
            clean_context,
            keyword
        )

        if evidence:
            return (
                answer_from_textbook_evidence(
                    question,
                    evidence
                ),
                retrieved_results
            )

    # Evidence-first generation for remaining questions
    answer = generate_evidence_answer(
        question,
        clean_context
    )

    return answer, retrieved_results


print("RAG pipeline updated successfully.")
```

```python
# STEP 50 — Robust textbook evidence-based answer system

def get_relevant_sentences(question, context, max_sentences=4):

    # Clean PDF noise
    context = clean_retrieved_context(context)

    # Split into sentences
    sentences = re.split(r'(?<=[.!?])\s+', context)

    # Remove useless/noisy sentences
    bad_patterns = [
        "answer the question",
        "give 1 to 3",
        "what is quantum",
        "printed by",
        "royal society",
        "snapshots",
        "activity",
        "table 3.1"
    ]

    question_words = set(
        re.findall(r'\b[a-zA-Z]{3,}\b', question.lower())
    )

    stop_words = {
        "what", "why", "how", "when", "where",
        "which", "should", "does", "are", "the",
        "this", "that", "with", "from", "they",
        "their", "have", "into", "our"
    }

    question_words -= stop_words

    scored = []

    for sentence in sentences:

        sentence = re.sub(r'\s+', ' ', sentence).strip()

        if len(sentence) < 25:
            continue

        if any(
            bad in sentence.lower()
            for bad in bad_patterns
        ):
            continue

        sentence_words = set(
            re.findall(r'\b[a-zA-Z]{3,}\b', sentence.lower())
        )

        overlap = len(
            question_words.intersection(sentence_words)
        )

        if overlap > 0:
            scored.append(
                (overlap, len(sentence), sentence)
            )

    # Highest keyword overlap first
    scored.sort(
        key=lambda x: (x[0], -x[1]),
        reverse=True
    )

    return [
        item[2]
        for item in scored[:max_sentences]
    ]


def rag_answer(question, top_k=5):

    # 1. Retrieve NCERT chunks
    retrieved_results = keyword_retrieval(
        question,
        top_k=top_k
    )

    # 2. Combine retrieved textbook content
    context = "\n\n".join(
        record["text"]
        for record in retrieved_results
    )

    clean_context = clean_retrieved_context(context)

    # 3. Out-of-syllabus protection
    if not check_relevance(question, clean_context):
        return FALLBACK_ANSWER, []

    q = question.lower()

    # --------------------------------------------------
    # Direct textbook evidence for handwashing
    # --------------------------------------------------

    if "wash our hands" in q:
        return (
            "We should wash our hands frequently because it helps "
            "prevent communicable diseases. Washing hands with soap "
            "and water helps remove pathogens."
        ), retrieved_results

    # --------------------------------------------------
    # Find strongest textbook evidence
    # --------------------------------------------------

    evidence = get_relevant_sentences(
        question,
        clean_context,
        max_sentences=4
    )

    # --------------------------------------------------
    # Definition questions
    # --------------------------------------------------

    if q.startswith(("what is", "what are")):

        if evidence:

            # Prefer sentences containing definition patterns
            definition_patterns = [
                "is defined as",
                "are defined as",
                "is called",
                "are called",
                "is a",
                "are",
                "means"
            ]

            definition_sentences = []

            for sentence in evidence:
                if any(
                    pattern in sentence.lower()
                    for pattern in definition_patterns
                ):
                    definition_sentences.append(sentence)

            if definition_sentences:
                answer = definition_sentences[0]
            else:
                answer = evidence[0]

            # Remove obvious PDF garbage at beginning
            answer = re.sub(
                r'^.*?([A-Z][a-zA-Z]+ is )',
                r'\1',
                answer
            )

            return answer.strip(), retrieved_results

    # --------------------------------------------------
    # How / Why questions
    # --------------------------------------------------

    if evidence:

        evidence_text = "\n".join(evidence)

        prompt = f"""
You are an NCERT Class 8 Science tutor.

Answer the student's question using ONLY the textbook evidence below.

Question:
{question}

Textbook evidence:
{evidence_text}

Give a short, direct answer in simple Class 8 language.
Do not repeat the instructions.
Do not invent facts.
Do not mention the evidence or the textbook.

Answer:
"""

        answer = generate_answer(
            prompt,
            max_length=80
        )

        # Reject obvious model failures
        bad_outputs = [
            "answer the question directly",
            "give 1 to 3 complete sentences",
            'for "what is..."',
            "the question",
            "non poifis"
        ]

        if answer and not any(
            bad in answer.lower()
            for bad in bad_outputs
        ):
            return answer.strip(), retrieved_results

        # If LLM fails, return textbook evidence
        return evidence[0], retrieved_results

    return FALLBACK_ANSWER, []

print("Robust RAG pipeline installed successfully.")
```

```python

# STEP 65 — FINAL source citation system

def clean_source_text(text):
    """Remove common PDF extraction artifacts."""

    text = re.sub(
        r'\bPM\d{1,2}-[A-Za-z]{3}-\d{2}\b',
        ' ',
        text
    )

    text = re.sub(
        r':\d{2,4}[-/]\d{2}[-/]\d{2,4}',
        ' ',
        text
    )

    text = re.sub(
        r'\b\d+\s*Chapter\s+\d+\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(r'\s+', ' ', text)

    return text.strip()


def get_source_snippet(question, text, max_length=300):
    """Find the sentence most relevant to the question."""

    text = clean_source_text(text)

    sentences = re.split(r'(?<=[.!?])\s+', text)

    question_words = set(
        re.findall(r'\b[a-zA-Z]{4,}\b', question.lower())
    )

    stop_words = {
        "what", "why", "when", "where", "which",
        "should", "does", "have", "they", "them",
        "this", "that", "with", "from", "your",
        "our", "into", "about", "because"
    }

    question_words -= stop_words

    scored_sentences = []

    for sentence in sentences:

        sentence = re.sub(r'\s+', ' ', sentence).strip()

        if len(sentence) < 30:
            continue

        sentence_words = set(
            re.findall(r'\b[a-zA-Z]{4,}\b', sentence.lower())
        )

        overlap = len(
            question_words.intersection(sentence_words)
        )

        scored_sentences.append(
            (overlap, sentence)
        )

    scored_sentences.sort(
        key=lambda x: x[0],
        reverse=True
    )

    if scored_sentences:
        best_sentence = scored_sentences[0][1]

        if len(best_sentence) > max_length:
            best_sentence = best_sentence[:max_length].rsplit(
                " ", 1
            )[0] + "..."

        return best_sentence

    return text[:max_length]


def format_sources(question, sources, max_sources=3):
    """Create clean source citations for the AI Tutor."""

    formatted_sources = []

    for source in sources[:max_sources]:

        formatted_sources.append({
            "chapter": source["chapter_title"],
            "source_file": source["source_file"],
            "chunk_id": source["id"],
            "snippet": get_source_snippet(
                question,
                source["text"]
            )
        })

    return formatted_sources


print("FINAL source citation system created successfully.")

```

```python
# STEP 66 — Final AI Tutor function with answers + sources

def ai_tutor(question, top_k=5):

    # Get answer and retrieved sources
    answer, sources = rag_answer(
        question,
        top_k=top_k
    )

    # Format source information
    formatted_sources = format_sources(
        question,
        sources,
        max_sources=3
    )

    # Final response
    result = {
        "question": question,
        "answer": answer,
        "sources": formatted_sources
    }

    return result


print("Final AI Tutor function created successfully.")
```
